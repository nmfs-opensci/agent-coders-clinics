```python
import boto3

# 1. Initialize the native AWS Bedrock Runtime client 
# (It automatically reads your AWS credentials from environment variables)
client = boto3.client("bedrock-runtime", region_name="us-east-1")

# 2. Define the Cross-Region Inference Profile ID for Claude Sonnet 4.6
model_id = "us.anthropic.claude-sonnet-4-6"

# 3. Structure the conversation
conversation = [
    {
        "role": "user",
        "content": [{"text": "Write a one-sentence bedtime story about a unicorn."}]
    }
]

try:
    # 4. Invoke the model using the standard Converse API
    response = client.converse(
        modelId=model_id,
        messages=conversation
    )
    
    # 5. Extract and print the text output
    output_text = response["output"]["message"]["content"][0]["text"]
    print(output_text)

except Exception as e:
    print(f"Error invoking model: {e}")
```
